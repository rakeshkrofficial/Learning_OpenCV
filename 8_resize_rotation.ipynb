```python
import cv2
import numpy as np
```

```python
img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/download_6.png")

re_img = cv2.resize(img,(500,500))

cv2.imshow("Original image",img)
cv2.imshow("resize image",re_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# rotate image
w,h = re_img.shape[0], re_img.shape[1]


m = cv2.getRotationMatrix2D((w/2,h/2),180,0.7)

new_img = cv2.warpAffine(re_img,m,(w,h))

cv2.imshow("original image",re_img)
cv2.imshow("rotate image",new_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
