```python
import cv2
import matplotlib.pyplot as plt

import numpy as np
```

```python
blank_w = np.ones((500,500,3),np.uint8) * 255
blank_b = np.zeros((500,500,3),np.uint8) * 255

hist = cv2.calcHist([blank_w],[0],None,[256],[0,256])

plt.plot(hist)
plt.show()

cv2.imshow("White",blank_w)
cv2.imshow("Black",blank_b)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
plt.figure(figsize = (10,10))

img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/download_6.png")
img = cv2.resize(img,(500,470))

b,g,r = cv2.split(img)

plt.hist(b.ravel(), bins=255, range=(0,255), color='b')
plt.hist(g.ravel(), bins=255, range=(0,255), color='r')
plt.hist(r.ravel(), bins=255,range=(0,255), color='g')

plt.show()

cv2.imshow("image",img)
cv2.imshow("Blue",b)
cv2.imshow("Green",g)
cv2.imshow("Red",r)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
