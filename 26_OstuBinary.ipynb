```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/noisy_leaf.jpg")
img = cv2.resize(img,(450,430))
img = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

hist = cv2.calcHist([img],[0],None,[255],[0,255])

plt.plot(hist)
plt.show()

_ , thr = cv2.threshold(img, 0, 255,cv2.THRESH_BINARY + cv2.THRESH_OTSU)

cv2.imshow("Leaf",img)
cv2.imshow("Threshold",thr)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
