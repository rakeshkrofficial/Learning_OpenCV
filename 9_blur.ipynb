```python
import cv2
import numpy as np
```

```python
img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/download_1.jpg")

re_img =cv2.resize(img,(300,300))

g = cv2.GaussianBlur(re_img,(7,7),0)
m = cv2.medianBlur(re_img,15)
b = cv2.bilateralFilter(re_img,25,100,100)

h = np.hstack((re_img,g,m,b))

cv2.imshow("Image",h)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
