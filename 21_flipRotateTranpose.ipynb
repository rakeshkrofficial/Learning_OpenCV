```python
import cv2
```

# Flip

```python
img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/download_7.jpg")
img = cv2.resize(img,(400,500))

img1 = cv2.flip(img,1) # y-axis
img2 = cv2.flip(img,0) # x-axis
img3 = cv2.flip(img,-1) # x-axis + y-axis

cv2.imshow("Original",img)
cv2.imshow(" across y-asix",img1)
cv2.imshow("across x-axis",img2)
cv2.imshow("across both x+y axis",img3)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Rotate

```python
img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/download_7.jpg")
img = cv2.resize(img,(400,500))

# img1 = cv2.rotate(img,cv2.ROTATE_180)  # rotate 180 degree
# img1 = cv2.rotate(img,cv2.ROTATE_90_CLOCKWISE)  # rotate 90 degree
img1 = cv2.rotate(img,cv2.ROTATE_90_COUNTERCLOCKWISE)


cv2.imshow("Original",img)
cv2.imshow("Rotate",img1)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Transpose

```python
img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/download_7.jpg")
img = cv2.resize(img,(400,500))

img1 = cv2.transpose(img)

cv2.imshow("Original",img)
cv2.imshow("Transpose image",img1)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
