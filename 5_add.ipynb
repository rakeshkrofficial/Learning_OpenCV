```python
img1 = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/mr.jpeg")
img2 = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/WhatsApp.jpeg")

img1 = cv2.resize(img1,(360,420))
img2 = cv2.resize(img2,(360,420))

new_img = cv2.addWeighted(img1,0.7,img2,0.5,1)

cv2.imshow("Add photo",new_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# substract
img1 = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/mr.jpeg")
img2 = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/WhatsApp.jpeg")

img1 = cv2.resize(img1,(360,420))
img2 = cv2.resize(img2,(360,420))

new_img = cv2.subtract(img2,img1)

cv2.imshow("Add photo",new_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
