```python
import cv2
import numpy as np
```

```python
# Write Text on Image
```

```python
img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/download_5.jpg")
img_get = cv2.resize(img,(750,550))



txt = cv2.putText(img=img_get,
    text ='Photographer',
    org = (75,150),
    fontFace = cv2.FONT_HERSHEY_DUPLEX,
    fontScale = 2,
    color = (203,192,255),
    thickness = 2,
    lineType = cv2.LINE_8,
    bottomLeftOrigin = False)  # img,text,org,fontFace, fontScale, color[, thickness[, lineType[,bottomLeftOrigin]]]]

cv2.imshow("Image",img_get)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
old_img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/download_6.PNG")
new_img = cv2.resize(old_img,(600,450))

text = cv2.putText(
    img = new_img,
    text =" Drow Rectangle On Face",
    org = (85,55),
    color =(255,255,255),
    fontFace =cv2.FONT_HERSHEY_DUPLEX,
    fontScale = 1,
    thickness = 2,
    lineType =cv2.LINE_8,
    
)

rectangle = cv2.rectangle(img = text ,
                         pt1 =(250,85),
                         pt2 =(360,309),
                         color =(0,255,255),
                         thickness=3,
                         lineType=5)

cv2.imshow("Image",rectangle)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Drow Cricle
```

```python
img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/download_6.PNG")
new_img = cv2.resize(img,(650,450))

circle_img = cv2.circle(
    img = new_img,
    radius = 95,
    center = (320,210),
    color =(0,255,0),
    thickness =2,
    lineType =16   # 1,4,8,16
    
)
cv2.imshow("Drow_circle",circle_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Drow ellipse
```

```python
img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/download_6.PNG")
new_img = cv2.resize(img,(650,450))

ellipse_drow = cv2.ellipse(
    img = new_img,
    center =(330,210),
    axes = (60,100),  # width , length
    angle = 5,
    startAngle = 0,
    endAngle = 360,
    color= (0,255,0),
    thickness = 2,
    lineType =16
)
    

cv2.imshow("Drow_ellipse",ellipse_drow)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Drow ellipse + text


img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/download_6.PNG")
new_img = cv2.resize(img,(650,450))

text = cv2.putText(
    img = new_img,
    text ='Teacher',
    org = (370,210),
    fontFace = 4,
    color = (255,255,255),
    fontScale = 1,
    thickness = 2,
    lineType =16,
    bottomLeftOrigin = False
)
ellipse_drow = cv2.ellipse(
    img = text,
    center =(330,210),
    axes = (60,100),  # width , length
    angle = 5,
    startAngle = 0,
    endAngle = 340,
    color= (0,255,0),
    thickness = 2,
    lineType =16
)

cv2.imshow("Drow_circle",ellipse_drow)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Draw Polygons on Image
```

```python
img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/download_7.jpg")
new_img = cv2.resize(img,(500,450))

polylines_drow = cv2.polylines(
    img = new_img,
    pts =[np.array([[150,150],[250,105],[380,125],[410,270],[170,270]])],
    isClosed = True,
    color = (0,255,0),
    thickness =2,
    lineType=16 )

    
cv2.imshow("Drow_ellipse",polylines_drow)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
