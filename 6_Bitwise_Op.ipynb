```python
import cv2
import numpy as np
```

```python
# And wise
```

```python
img1 = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/binary.png")
img2 = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/binary_1.png")

img1 = cv2.resize(img1 , (350,400))
img2  = cv2.resize(img2, (350,400))

new_img = cv2.bitwise_and(img1,img2)

h= np.hstack((img1,img2,new_img))

cv2.imshow("Binary Image", h)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img1 = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/binary.png")
img2 = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/binary_1.png")

img1 = cv2.resize(img1 , (300,300))
img2  = cv2.resize(img2, (300,300))

new_img1 = cv2.bitwise_not(img1)
new_img2 = cv2.bitwise_not(img2)

h1= np.hstack((img1,new_img1))
h2 = np.hstack((img2,new_img2))

v = np.vstack((h1,h2))

cv2.imshow("Binary Image", v)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
