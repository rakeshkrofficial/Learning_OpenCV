```python
import cv2
import numpy as np
```

```python
# erosion

img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/binary_3.png")
img = cv2.resize(img,(400,400))

m = np.ones((5,5), np.int8)

er = cv2.erode(img , m,iterations=1)

cv2.imshow("erode",er)
cv2.imshow("image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Dilation

img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/binary_3.png")
img = cv2.resize(img,(400,400))

m = np.ones((5,5), np.int8)

di = cv2.dilate(img,m,iterations = 2)

cv2.imshow("dilate",di)
cv2.imshow("image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# opening

img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/binary_3.png")
img = cv2.resize(img,(400,400))

k = np.ones((5,5),np.int8)

op = cv2.morphologyEx(img,cv2.MORPH_OPEN,k,iterations = 3)

cv2.imshow("Image",img)
cv2.imshow("Opening image",op)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# closing

img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/binary_4.png")
img = cv2.resize(img,(400,400))

k = np.ones((5,5),np.int8)

cl = cv2.morphologyEx(img,cv2.MORPH_CLOSE,k,iterations= 1)

cv2.imshow("Image",img)
cv2.imshow("closing image",cl)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Morphological Gradient

img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/binary_3.png")
img = cv2.resize(img,(400,400))

k = np.ones((5,5),np.int8)

gr = cv2.morphologyEx(img,cv2.MORPH_GRADIENT,k,iterations = 1)

cv2.imshow("image",img)
cv2.imshow("gradient",gr)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Tophat

img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/binary_5.png")
img = cv2.resize(img,(400,400))

k = np.ones((29,29),np.int8)

tp = cv2.morphologyEx(img,cv2.MORPH_TOPHAT,k,iterations=1)

cv2.imshow("Image",img)
cv2.imshow("tophat",tp)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Blackhat

img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/binary_7.png")
img = cv2.resize(img,(400,400))

k = np.ones((50,50),np.int8)

tp = cv2.morphologyEx(img,cv2.MORPH_BLACKHAT,k,iterations=1)

cv2.imshow("Image",img)
cv2.imshow("tophat",tp)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
