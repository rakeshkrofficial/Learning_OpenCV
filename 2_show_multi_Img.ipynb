```python
# vertical
```

```python
img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/download_1.jpg")

re_size = cv2.resize(img,(300,300))

v = np.vstack((re_size,re_size))
cv2.imshow("Horizontally",v)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# horizontal + vertical
```

```python
img = cv2.imread("C:/Users/rakes/Documents/Machine Learning/Images/download_1.jpg")

re_size = cv2.resize(img,(300,300))

h = np.hstack((re_size,re_size,re_size))
v = np.vstack((h,h))

cv2.imshow("Horizontally",v)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
